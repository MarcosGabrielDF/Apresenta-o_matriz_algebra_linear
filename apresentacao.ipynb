## O que é o NumPy?

O **NumPy** (*Numerical Python*) é a biblioteca fundamental para computação numérica e álgebra linear em Python. Em vez de trabalhar com listas comuns do Python — que são lentas para grandes volumes numéricos e não entendem operações matemáticas diretas em blocos —, o NumPy introduz o **`ndarray`** (um array multidimensional homogêneo e de tamanho fixo).

### Por que usar NumPy em Álgebra Linear e Ciência de Dados?
* **Desempenho:** suas operações são executadas por rotinas compiladas em C/Fortran sob os panos, sendo muito mais rápidas que loops `for` tradicionais em Python.
* **Vetorização e broadcasting:** você aplica operações como `+`, `-` e multiplicação por escalar diretamente sobre arrays inteiros, de forma limpa e legível.
* **Base do ecossistema:** bibliotecas como `pandas`, `scipy`, `scikit-learn` e `matplotlib` constroem suas estruturas e cálculos em cima ou em integração direta com arrays NumPy.

### O que observar ao trabalhar com matrizes no NumPy
* **Formato (`shape`):** indica as dimensões do array — por exemplo, `(2, 2)` significa 2 linhas e 2 colunas.
* **Tipo dos dados (`dtype`):** define como os números são armazenados (como `float64` ou `int64`), o que impacta precisão e operações.
* **Multiplicação elemento a elemento vs. matricial:**
  * `A * B` multiplica posição por posição (exige mesmas dimensões ou *broadcasting*).
  * `A @ B` (ou `np.matmul(A, B)`) executa o **produto matricial** da álgebra linear.

# Operações com Matrizes usando NumPy

Este notebook demonstra quatro operações fundamentais de álgebra linear com **NumPy**:
1. **Soma de matrizes** (`+`)
2. **Subtração de matrizes** (`-`)
3. **Multiplicação por escalar** (`*`)
4. **Multiplicação de matrizes** (`@` / `np.matmul`)

## Configurações básicas.

```python
import numpy as np

# Definimos duas matrizes 2x2 com tipo float para evitar truncamento em cálculos futuros
matriz_a = np.array([
    [1, 2],
    [3, 4]
], dtype=float)

matriz_b = np.array([
    [5, 6],
    [7, 8]
], dtype=float)

print("Matriz A:\n", matriz_a)
print("\nMatriz B:\n", matriz_b)
```

## Soma e subtração 

```python
# Soma elemento por elemento: exige que A e B tenham o mesmo shape (2, 2)
soma = matriz_a + matriz_b

# Subtração elemento por elemento: exige que A e B tenham o mesmo shape (2, 2)
subtracao = matriz_a - matriz_b

print("Soma (A + B):\n", soma)
print("\nSubtração (A - B):\n", subtracao)
```

## Multiplicar a matriz por um escalar

```python
# Multiplica cada elemento de A pelo escalar escolhido
escalar = 3
mult_escalar = escalar * matriz_a

print(f"Multiplicação por escalar ({escalar} * A):\n", mult_escalar)
```

## Multiplicação de matrizes

```python
# Produto matricial (linha x coluna): use '@' ou np.matmul()
# Nota: '*' faria multiplicação elemento a elemento, não produto matricial.
mult_matriz = matriz_a @ matriz_b

# Demonstrando que a ordem importa (A @ B != B @ A em geral)
mult_matriz_inversa = matriz_b @ matriz_a

# Multiplicação pela matriz identidade (I):
# np.eye(2) cria uma matriz identidade 2x2 (1 na diagonal principal, 0 no resto)
identidade_2x2 = np.eye(2)

# Ao contrário de matrizes comuns, multiplicar pela identidade preserva A
# tanto à direita (A @ I) quanto à esquerda (I @ A)
mult_identidade_dir = matriz_a @ identidade_2x2
mult_identidade_esq = identidade_2x2 @ matriz_a

print("Multiplicação de matrizes (A @ B):\n", mult_matriz)
print("\nMultiplicação na ordem inversa (B @ A):\n", mult_matriz_inversa)
print("\nMatriz Identidade (I 2x2):\n", identidade_2x2)
print("\nMultiplicação por identidade (A @ I):\n", mult_identidade_dir)
print("\nMultiplicação por identidade na outra ordem (I @ A):\n", mult_identidade_esq)
```
